Q3. What advantages do stored procedures offer compared to writing raw SQL queries repeatedly

**Ans-**

Stored Procedures are pre-written SQL programs stored in the database and executed when needed.

**Advantages over repeating raw SQL queries:**

*   Reusability: Write once, use many times.

*   Better Performance: Compiled and optimized by the database.
*   Improved Security: Users can execute procedures without direct table access.


*   Easy Maintenance: Changes are made in one place.



Q4. What is the purpose of triggers in a database? Mention one use case where a trigger is essential.

**Ans-**

Triggers are special database programs that automatically execute when certain events occur, such as INSERT, UPDATE, or DELETE.

**Use Case:**

An audit log trigger can automatically record changes made to important tables whenever data is updated.

Q5. Explain the need for data modelling and normalization when designing a database.

**Ans-**

**Data Modelling** helps organize data and define relationships between tables, ensuring an efficient database structure.

**Normalization** reduces data redundancy and improves data consistency by organizing data into related tables.

**Benefits:**

*   Minimizes duplicate data
*   Maintains data accuracy


*   Improves database efficiency
*   Makes updates and maintenance easier
